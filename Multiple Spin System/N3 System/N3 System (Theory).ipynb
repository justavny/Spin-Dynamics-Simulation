# N=3 System

### 1. Why exact evolution breaks down at N = 3

For a 3-spin open chain with only nearest-neighbor coupling, the Hamiltonian is

$$
\hat H = -J_x\!\left[\hat S_1^x\hat S_2^x + \hat S_2^x\hat S_3^x\right]
        -J_y\!\left[\hat S_1^y\hat S_2^y + \hat S_2^y\hat S_3^y\right]
        -J_z\!\left[\hat S_1^z\hat S_2^z + \hat S_2^z\hat S_3^z\right].
$$

Writing $\hat H = \hat h_{12} + \hat h_{23}$, the two pieces do not commute in general ($[\hat h_{12},\hat h_{23}]\neq 0$, since they share qubit 2), so $e^{-i\hat H t} \;\neq\; e^{-i\hat h_{12}t}\,e^{-i\hat h_{23}t}.$

Unless the specific couplings happen to make the chain exactly solvable/integrable, there is no simple closed-form expression for $\hat U(t)$ built out of elementary gates. We therefore need an approximate method, the Lie–Trotter decomposition.

### 2. The Lie–Trotter product formula

For non-commuting operators $\hat A,\hat B$, the Trotter product formula states

$
e^{-i(\hat A+\hat B)t} = \lim_{n\to\infty}\left(e^{-i\hat A t/n}\,e^{-i\hat B t/n}\right)^{n}.
$

Splitting a time interval into $n$ sub-steps and applying each generator separately converges to the exact evolution as $n\to\infty$. For finite $n$, the leading error (from the Baker–Campbell–Hausdorff expansion) scales as

$$
\left\| U_{\text{exact}}(t) - U_{\text{Trotter}}^{(n)}(t)\right\| = O\!\left(\frac{t^2}{n}\right),
$$

so more Trotter steps improve accuracy. On real quantum hardware this creates a genuine trade-off: each Trotter step of the 3-spin chain costs two applications of the 3-CNOT two-qubit gate (one for $\hat h_{12}$, one for $\hat h_{23}$), so increasing $n$ increases circuit depth, and CNOT gates are the dominant source of hardware error. 

### 3. Trotterizing the N = 3 open chain

Define the two-spin blocks acting on the full 3-spin Hilbert space ($2^3=8$ dimensional):

$$
\hat h_{12} = -J_x\hat S_1^x\hat S_2^x - J_y\hat S_1^y\hat S_2^y - J_z\hat S_1^z\hat S_2^z \;\otimes\; \hat I_3,
\qquad
\hat h_{23} = \hat I_1 \;\otimes\; \left(-J_x\hat S_2^x\hat S_3^x - J_y\hat S_2^y\hat S_3^y - J_z\hat S_2^z\hat S_3^z\right).
$$

Each factor $e^{-i\hat h_{12}(t/n)}$ or $e^{-i\hat h_{23}(t/n)}$ is exactly the same two-qubit gate used for the N = 2 case (Lancaster & Allen's $\hat N(\alpha,\beta,\gamma)$, 3 CNOTs), just acting on a different pair of qubits within the 3-qubit register with identity on the spectator qubit. The full approximate propagator is

$
\hat U(t) \;\approx\; \left[e^{-i\hat h_{12}(t/n)}\, e^{-i\hat h_{23}(t/n)}\right]^{n}.
$
