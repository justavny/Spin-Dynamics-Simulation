# Multiple Spin Systems

### 1. Introduction

A single spin-${\frac{1}{2}}$ particle lives in a two-dimensional Hilbert space and maps naturally onto a single qubit, as seen in the previous notebook. Once a second spin is introduced, the state of the combined system no longer lives in a simple sum of two two-dimensional spaces, it exists in their tensor product.

For two spins, the joint Hilbert space is spanned by the basis ${\{\,|{+}{+}\rangle,\;|{+}{-}\rangle,\;|{-}{+}\rangle,\;|{-}{-}\rangle\,\}}$ which is four-dimensional. In general, a system of ${N}$ spins requires a Hilbert space of dimension ${2^N}$, which is why simulating even modestly-sized spin systems becomes classically intractable, and why quantum computers, whose qubit registers grow the same way, are naturally suited to the task.

### 2. Product States vs. Entangled States

Some two-spin states can be written as a simple product of individual spin states,

$${
|\psi\rangle = |\psi_1\rangle \otimes |\psi_2\rangle
}$$

These are called product states, and they contain no more information than the two individual spins considered separately. However, not every state in the joint Hilbert space can be factored this way. States such as

$${
|\Phi^{\pm}\rangle = \frac{1}{\sqrt{2}}\big(|{+}{+}\rangle \pm |{-}{-}\rangle\big), \qquad
|\Psi^{\pm}\rangle = \frac{1}{\sqrt{2}}\big(|{+}{-}\rangle \pm |{-}{+}\rangle\big)
}$$

cannot be written as a single tensor product of two one-spin states. These are the Bell states, and a system prepared in one of them is said to be entangled. Measuring one spin instantaneously determines the outcome of measuring the other, regardless of the spatial separation between them, which cannot be done in classical systems, as it would violate the principle of locality.

### 3. Why N = 2 is Special

For time-independent Hamiltonians acting on ${N \leq 2}$ spins, the time-evolution operator ${U(t) = e^{-iHt}}$ can be constructed exactly as a quantum circuit (no approximation is required). Once ${N \geq 3}$, the interaction terms no longer commute in a way that allows an exact, closed-form circuit, and one must resort to the Lie–Trotter decomposition to approximate the evolution. This distinction exact evolution for small systems, approximate (Trotterized) evolution for larger ones is the organizing idea behind the rest of this repository.

### Key Idea

A second spin does not just double the system size, it opens up an entirely new resource, entanglement, with correlations that cannot be explained by any classical, local model. Two-spin systems are the smallest setting in which this can be demonstrated, both theoretically and experimentally on real quantum hardware.
