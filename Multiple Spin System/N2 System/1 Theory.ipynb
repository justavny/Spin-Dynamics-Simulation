# N=2 System

### 1. The Two-Spin Hamiltonian

Following the treatment of interacting spins used to model exact two-spin dynamics on a quantum computer, we consider two spin-${\frac{1}{2}}$ particles coupled through an isotropic exchange interaction,

$${
\hat H_2 = J\, \hat{\bf S}_1 \cdot \hat{\bf S}_2
= J\left(\hat S_1^x \hat S_2^x + \hat S_1^y \hat S_2^y + \hat S_1^z \hat S_2^z\right)
}$$

where ${J}$ sets the strength of the coupling between the two spins. This is the same form of interaction responsible for the energy splitting between singlet and triplet configurations in atomic and molecular physics.

### 2. Total Spin and Conservation Laws

Because ${\hat H_2}$ depends only on the dot product ${\hat{\bf S}_1 \cdot \hat{\bf S}_2}$, it commutes with the total spin operator

$${
\hat{\bf S} = \hat{\bf S}_1 + \hat{\bf S}_2, \qquad [\hat H_2,\, \hat{\bf S}^2] = 0, \qquad [\hat H_2,\, \hat S^z] = 0
}$$

This is a direct example of the connection between symmetries and conservation laws: rotational symmetry of the exchange interaction guarantees that total spin and its z-projection are conserved quantities. The eigenstates of ${\hat H_2}$ can therefore be organized by total spin quantum number.

### 3. Singlet and Triplet Eigenstates

Diagonalizing ${\hat H_2}$ in the two-spin basis gives four eigenstates:

$${
|\text{triplet}\rangle:\quad
|{+}{+}\rangle,\;\;
\frac{1}{\sqrt{2}}\big(|{+}{-}\rangle + |{-}{+}\rangle\big),\;\;
|{-}{-}\rangle
\qquad (S = 1)
}$$

$${
|\text{singlet}\rangle:\quad
\frac{1}{\sqrt{2}}\big(|{+}{-}\rangle - |{-}{+}\rangle\big)
\qquad (S = 0)
}$$

The singlet is the unique ${S=0}$ state and is maximally entangled. It cannot be written as a product of single-spin states. The triplet states are ${S=1}$ and are degenerate in energy with respect to each other only in the absence of an external field.

### 4. Time Evolution

Since ${\hat H_2}$ is time-independent, evolution is again given by ${U(t) = e^{-i\hat H_2 t/\hbar}}$. Unlike the single-spin case, ${\hat H_2}$ does not reduce to a single-qubit rotation, instead it acts non-trivially across both qubits. However, because ${N=2}$, the exact evolution operator can still be decomposed into a finite sequence of standard two-qubit gates (e.g. combinations of ${CNOT}$, single-qubit rotations, and a controlled-phase-like entangling gate), allowing the dynamics to be implemented exactly, without any Trotter approximation.


### Key Idea

The two-spin exchange Hamiltonian conserves total spin as a direct consequence of its rotational symmetry, producing a singlet and a triplet manifold. The singlet state is maximally entangled, and — crucially for what follows — its correlations cannot be reproduced by any classical, local model. Because ${N=2}$ is small enough for exact circuit synthesis, both the entangling dynamics and its consequences can be verified precisely, without approximation, on real quantum hardware.
