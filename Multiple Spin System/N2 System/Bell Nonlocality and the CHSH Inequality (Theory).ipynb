# Bell Nonlocality and the CHSH Inequality

### 1. Motivation

The singlet state derived in the previous notebook, ${|\Psi^-\rangle = \frac{1}{\sqrt{2}}\big(|{+}{-}\rangle - |{-}{+}\rangle\big)}$ is entangled, but entanglement alone does not automatically imply a violation of classical physics. What makes the singlet special is that measurements performed on its two spins, along independently chosen axes, produce correlations that no local hidden-variable theory can reproduce. This is the content of Bell's theorem, and the CHSH inequality gives a way to test it experimentally.

### 2. Local Hidden Variable Assumption

Suppose each spin carries some (possibly hidden) instruction set ${\lambda}$ determining the outcome of any measurement made on it, independent of what is measured on the other, distant spin. Under this assumption, the correlation between measurements along axes ${\hat a}$ and ${\hat b}$ is bounded by the classical CHSH inequality,

$${
S = E(\hat a,\hat b) - E(\hat a,\hat b') + E(\hat a',\hat b) + E(\hat a',\hat b') , \qquad |S| \leq 2
}$$

where ${E(\hat a,\hat b)}$ is the expectation value of the product of measurement outcomes (${\pm 1}$) when spin 1 is measured along ${\hat a}$ and spin 2 along ${\hat b}$.

### 3. Quantum Prediction for the Singlet

For the singlet state, quantum mechanics predicts

$${
E(\hat a, \hat b) = \langle \Psi^-|\, (\hat{\boldsymbol\sigma}_1\!\cdot\hat a)(\hat{\boldsymbol\sigma}_2\!\cdot\hat b)\,|\Psi^-\rangle = -\,\hat a \cdot \hat b
}$$

If the two measurement axes are separated by an angle ${\theta}$, this becomes ${E(\theta) = -\cos\theta}$.

### 4. Choosing Optimal Angles

Substituting into the CHSH combination with the standard choice of angles ${\hat a = 0^\circ}$, ${\hat a' = 45^\circ}$, ${\hat b = 22.5^\circ}$, ${\hat b' = 67.5^\circ}$ gives

$${
S = -\cos(22.5^\circ) + \cos(67.5^\circ) - \cos(22.5^\circ) - \cos(67.5^\circ)
}$$

which simplifies to

$${
S = -2\sqrt{2} \approx -2.828
}$$

Taking the magnitude, ${|S| = 2\sqrt{2}}$, exceeding the classical bound of 2. This is the Tsirelson bound, the maximum CHSH violation allowed by quantum mechanics.


### Key Idea

The singlet state's correlations, measured along four suitably chosen axes, produce a CHSH value of ${2\sqrt2}$ which is a violation of the classical bound of 2 that cannot be explained by any local hidden-variable model.
